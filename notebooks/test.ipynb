```python
import xarray as xr
from dask.diagnostics import ProgressBar

# Open lazily
data = xr.open_dataset(
    "https://nyu1.osn.mghpcc.org/m2lines-pubs/Samudra/OM4",
    engine="zarr",
    chunks="auto"
)

# Subset
subset = data.sel(time=slice("2014-10-10", "2022-12-24"))

# Optional rechunk
subset = subset.chunk({"time": 10})

# Progress bar
with ProgressBar():
    subset.to_zarr("data", mode="w")
```

```python
import xarray as xr

# Download statistics
means = xr.open_dataset("https://nyu1.osn.mghpcc.org/m2lines-pubs/Samudra/OM4_means", engine='zarr', chunks={})
means.to_zarr("data/means.zarr")

stds = xr.open_dataset("https://nyu1.osn.mghpcc.org/m2lines-pubs/Samudra/OM4_stds", engine='zarr', chunks={})
stds.to_zarr("data/stds.zarr")
```
